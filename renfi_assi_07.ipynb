```python
# Accept number of time steps
n = int(input("Enter the number of time steps: "))

# Accept reward for each time step
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Accept discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Calculate discounted return
discounted_return = 0

for i in range(n):
    discounted_return += rewards[i] * (gamma ** i)

# Display results
print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

```python
%%writefile grid_env_det.py

# 2 x 3 Grid World MDP

# State Space
STATES = ["S0", "S1", "S3", "S4", "G"]

# Action Space
ACTIONS = ["Up", "Down", "Left", "Right"]

# Discount Factor
GAMMA = 0.9

# Transition Function
TRANSITIONS = {
    "S0": {
        "Up": "S0",
        "Down": "S0",
        "Left": "S0",
        "Right": "S1"
    },

    "S1": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S0",
        "Right": "G"
    },

    "S3": {
        "Up": "S1",
        "Down": "S3",
        "Left": "S3",
        "Right": "S4"
    },

    "S4": {
        "Up": "G",
        "Down": "S4",
        "Left": "S3",
        "Right": "S4"
    },

    "G": {
        "Up": "G",
        "Down": "G",
        "Left": "G",
        "Right": "G"
    }
}

# Reward Function
def get_reward(current_state, action):
    # Goal state
    if current_state == "G":
        return 10

    # Find next state
    next_state = TRANSITIONS[current_state][action]

    # If action reaches goal
    if next_state == "G":
        return 10

    # Obstacle movements
    obstacle_moves = {
        ("S0", "Down"),
        ("S3", "Left")
    }

    if (current_state, action) in obstacle_moves:
        return -5

    # Normal movement
    return -1


# Transition Function
def transition(current_state, action):
    if current_state not in STATES:
        return "Invalid state"

    if action not in ACTIONS:
        return "Invalid action"

    return TRANSITIONS[current_state][action]
```

```python
# Import the Grid World environment
import grid_env_det as env

# Display State Space
print("State Space:", env.STATES)

# Display Action Space
print("Action Space:", env.ACTIONS)

# Display Discount Factor
print("Discount Factor (gamma):", env.GAMMA)

# Display Reward Function
print("\nReward Function:")
print("Normal movement  = -1")
print("Goal state        = +10")
print("Obstacle movement = -5")

# Take input from user
current_state = input("\nEnter current state (S0, S1, S3, S4, G): ")
action = input("Enter action (Up, Down, Left, Right): ")

# Find next state
next_state = env.transition(current_state, action)

# Find reward
reward = env.get_reward(current_state, action)

# Display result
print("\nCurrent State:", current_state)
print("Action:", action)
print("Next State:", next_state)
print("Reward:", reward)
```

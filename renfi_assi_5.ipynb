```python
import gymnasium as gym


def run_random_agent(environment_name, episodes=5):
    env = gym.make(environment_name)

    rewards = []

    print("\nEnvironment:", environment_name)

    for episode in range(episodes):
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        while not (terminated or truncated):
            # Select random action
            action = env.action_space.sample()

            # Perform action
            next_state, reward, terminated, truncated, info = env.step(action)

            total_reward += reward
            state = next_state

        rewards.append(total_reward)

        print("Episode", episode + 1, "Reward:", total_reward)

    # Calculate total and average reward
    total = sum(rewards)
    average = total / len(rewards)

    print("Total Reward:", total)
    print("Average Reward:", average)

    env.close()


# Run all three environments
run_random_agent("CartPole-v1")
run_random_agent("MountainCar-v0")
run_random_agent("FrozenLake-v1")
```

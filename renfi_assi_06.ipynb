```python


!pip install -q -U gymnasium

import gymnasium as gym


def random_agent(environment_name, number_of_episodes=5):

    env = gym.make(environment_name)

    print("\n" + "=" * 60)
    print("Environment:", environment_name)
    print("=" * 60)

    for episode in range(1, number_of_episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0
        terminated = False
        truncated = False

        # State -> Action -> Reward -> Next State
        while not (terminated or truncated):

            # Select random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Update total reward
            total_reward += reward
            steps += 1

            # Move to next state
            state = next_state

        # Calculate average reward
        average_reward = total_reward / steps

        print("Episode:", episode)
        print("Total Reward:", total_reward)
        print("Number of Steps:", steps)
        print("Average Reward:", round(average_reward, 2))
        print("-" * 60)

    env.close()


random_agent("CliffWalking-v1", 5)



random_agent("Taxi-v4", 5)
```

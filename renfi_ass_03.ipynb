```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
total_plays = 10

rewards = {}

for machine in machines:
    print(f"\nEnter the 4 rewards for Machine {machine}:")
    rewards[machine] = []

    for i in range(4):
        value = float(input(f"Reward {i+1} for Machine {machine}: "))
        rewards[machine].append(value)

reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}


N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': 0,
    'B': 0,
    'C': 0
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

total_reward = 0

random.seed(42)

print("\n" + "=" * 55)
print("STARTING MULTI-ARMED BANDIT")
print("=" * 55)

for play in range(1, total_plays + 1):

    available = [
        m for m in machines
        if reward_index[m] < len(rewards[m])
    ]

    if len(available) == 0:
        print("\nNo more rewards available.")
        break
    max_q = max(Q[m] for m in available)

    best_machines = [
        m for m in available
        if Q[m] == max_q
    ]

    highest_machine = best_machines[0]

    random_number = random.random()



    if random_number < epsilon:

        exploration_choices = [
            m for m in available
            if m != highest_machine
        ]
        if len(exploration_choices) == 0:
            selected = highest_machine
        else:
            selected = random.choice(exploration_choices)

        method = "Exploration"

    else:


        selected = highest_machine
        method = "Exploitation"


    index = reward_index[selected]
    reward = rewards[selected][index]

    reward_index[selected] += 1

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    total_reward += reward

    print(f"\nPlay {play}")
    print(f"Random number = {random_number:.4f}")
    print(f"Method = {method}")
    print(f"Selected Machine = {selected}")
    print(f"Reward received = {reward}")

    print(f"N = {N}")
    print(f"R = {R}")

    print("Q values:")
    print(f"A = {Q['A']:.2f}")
    print(f"B = {Q['B']:.2f}")
    print(f"C = {Q['C']:.2f}")


print("\n" + "=" * 55)
print("FINAL RESULTS")
print("=" * 55)

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)
```

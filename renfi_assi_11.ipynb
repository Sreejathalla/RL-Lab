```python

gamma = 0.9
theta = 0.000001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

normal_reward = -1
wall_reward = -1
goal_reward = 100

def transition(state, action):

    if state == 'S0':

        transitions = {
            'Up': ('S0', normal_reward),
            'Down': ('S1', normal_reward),
            'Left': ('S0', normal_reward),
            'Right': ('S0', wall_reward)
        }

    elif state == 'S1':

        transitions = {
            'Up': ('S0', normal_reward),
            'Down': ('S1', normal_reward),
            'Left': ('S1', wall_reward),
            'Right': ('G', goal_reward)
        }

    return transitions[action]

policy = {
    'S0': 'Up',
    'S1': 'Up'
}

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

print("INITIAL POLICY")
print("S0 =", policy['S0'])
print("S1 =", policy['S1'])

print("\nINITIAL STATE VALUES")
print(V)


# 4. Policy Evaluation

def policy_evaluation(policy, V):

    iteration = 0

    while True:

        delta = 0
        new_V = V.copy()

        for state in states:

            action = policy[state]

            next_state, reward = transition(state, action)

            value = reward + gamma * V[next_state]

            new_V[state] = value

            delta = max(delta, abs(value - V[state]))

        V.update(new_V)
        iteration += 1

        if delta < theta:
            break

    print("\nPolicy Evaluation Iterations:", iteration)

    for state in states:
        print(f"V({state}) = {V[state]:.4f}")

    return V


# 5. Calculate Action Values

def calculate_action_values(state, V):

    Q = {}

    for action in actions:

        next_state, reward = transition(state, action)

        Q[action] = reward + gamma * V[next_state]

    return Q


# 6. Policy Improvement

def policy_improvement(policy, V):

    stable = True

    print("\nACTION VALUE CALCULATIONS")

    for state in states:

        old_action = policy[state]

        Q = calculate_action_values(state, V)

        print("\nFor", state)

        for action in actions:
            print(f"Q({state}, {action}) = {Q[action]:.2f}")

        best_action = max(actions, key=lambda a: Q[a])

        policy[state] = best_action

        print("Updated Action:", best_action)

        if old_action != best_action:
            stable = False

    return policy, stable


# 7. Policy Iteration

iteration = 1

while True:

    print("\n" + "=" * 50)
    print("POLICY ITERATION", iteration)
    print("=" * 50)

    # Policy Evaluation
    V = policy_evaluation(policy, V)

    # Policy Improvement
    policy, stable = policy_improvement(policy, V)

    print("\nUPDATED POLICY")
    print("S0 =", policy['S0'])
    print("S1 =", policy['S1'])

    if stable:
        print("\nPOLICY IS STABLE")
        break

    iteration += 1


# 8. Final Optimal State Values

print("\n" + "=" * 50)
print("FINAL OPTIMAL RESULTS")
print("=" * 50)

print("\nOPTIMAL STATE VALUES")

for state in states:
    print(f"V*({state}) = {V[state]:.2f}")

print("V*(G) = 0.00")


# 9. Final Optimal Policy

print("\nOPTIMAL POLICY")

for state in states:
    print(f"{state} -> {policy[state]}")

print("G -> Terminal")


# 10. Final Action Value Table

print("\nFINAL ACTION VALUE TABLE")

for state in states:

    Q = calculate_action_values(state, V)

    print("\nState:", state)

    for action in actions:
        print(f"Q*({state}, {action}) = {Q[action]:.2f}")


# 11. Bellman Optimality Verification

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:

    Q = calculate_action_values(state, V)

    max_Q = max(Q.values())

    print(f"\nV*({state}) = {V[state]:.2f}")

    print(f"Max Q*({state}, a) = {max_Q:.2f}")

    if abs(V[state] - max_Q) < 0.001:
        print("Bellman Equation Verified!")


# 12. Optimal Path

print("\nOPTIMAL PATH")
print("S0 -> S1 -> G")
```

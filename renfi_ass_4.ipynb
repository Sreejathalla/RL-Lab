```python
import math

machines = ['A', 'B', 'C']

M = 3
N_play = 10

N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
UCB = [0.0, 0.0, 0.0]

total_reward = 0

print("UCB Multi-Armed Bandit")
print("----------------------")


for i in range(M):

    print("\nInitial Test")
    print("Play with Machine", machines[i])

    reward = float(input("Enter reward: "))


    N[i] = N[i] + 1

    R[i] = R[i] + reward

    Q[i] = Q[i] + (reward - Q[i]) / N[i]

    total_reward = total_reward + reward


for play in range(4, N_play + 1):

    for i in range(M):
        UCB[i] = Q[i] + 2 * math.sqrt(
            math.log(play) / N[i]
        )

    selected = UCB.index(max(UCB))

    print("\nPlay", play)

    print("UCB Values:")
    for i in range(M):
        print(machines[i], "=", round(UCB[i], 4))

    print("Selected Machine:", machines[selected])

    reward = float(input(
        "Enter reward for Machine " + machines[selected] + ": "
    ))
    N[selected] = N[selected] + 1


    R[selected] = R[selected] + reward


    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]


    total_reward = total_reward + reward


for i in range(M):
    UCB[i] = Q[i] + 2 * math.sqrt(
        math.log(N_play) / N[i]
    )

print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

for i in range(M):
    print(
        machines[i], "\t",
        N[i], "\t",
        R[i], "\t",
        round(UCB[i], 4), "\t\t",
        round(Q[i], 4)
    )

print("\nTotal Reward =", total_reward)
```

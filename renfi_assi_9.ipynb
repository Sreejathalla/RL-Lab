```python
gamma = 0.9
theta = 0.000001

# Initialize state values
V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0,
    'Wall': 0.0
}

# Policy probabilities
policy = {
    'S0': {'Down': 0.9, 'Right': 0.1},
    'S1': {'Right': 0.9, 'Up': 0.1}
}

# State transitions
transitions = {
    ('S0', 'Down'): ('S1', -1),
    ('S0', 'Right'): ('Wall', -1),
    ('S1', 'Right'): ('G', 100),
    ('S1', 'Up'): ('S0', -1)
}

# Policy Evaluation
iteration = 0

while True:
    delta = 0
    new_V = V.copy()

    # Update S0
    new_V['S0'] = (
        0.9 * (-1 + gamma * V['S1']) +
        0.1 * (-1 + gamma * V['Wall'])
    )

    # Update S1
    new_V['S1'] = (
        0.9 * (100 + gamma * V['G']) +
        0.1 * (-1 + gamma * V['S0'])
    )

    delta = max(
        abs(new_V['S0'] - V['S0']),
        abs(new_V['S1'] - V['S1'])
    )

    V = new_V
    iteration += 1

    print("Iteration", iteration,
          "S0 =", round(V['S0'], 6),
          "S1 =", round(V['S1'], 6))

    if delta < theta:
        break

# Calculate Action Values
Q = {}

Q[('S0', 'Down')] = -1 + gamma * V['S1']

Q[('S0', 'Right')] = -1 + gamma * V['Wall']

Q[('S1', 'Right')] = 100 + gamma * V['G']

Q[('S1', 'Up')] = -1 + gamma * V['S0']

# Display results
print("\nNumber of iterations:", iteration)

print("\nFinal State Values:")
print("V(S0) =", round(V['S0'], 2))
print("V(S1) =", round(V['S1'], 2))

print("\nAction Values:")
for key, value in Q.items():
    print("Q", key, "=", round(value, 2))

# Policy Consistency Check
print("\nPolicy Consistency Check:")
print("V(S0) =", round(V['S0'], 2))
print("Q(S0, Down) =", round(Q[('S0', 'Down')], 2))

print("V(S1) =", round(V['S1'], 2))
print("Q(S1, Right) =", round(Q[('S1', 'Right')], 2))
```

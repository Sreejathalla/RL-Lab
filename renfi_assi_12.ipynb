```python
# ============================================================
# ASSIGNMENT 12: VALUE ITERATION IN DETERMINISTIC GRID WORLD
# ============================================================

# 1. Define Parameters

gamma = 0.9
theta = 0.0001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

normal_reward = -1
wall_reward = -1
goal_reward = 100

# Grid World:
# S0 | WALL
# S1 | G


# 2. Define Transition Function

def transition(state, action):

    if state == 'S0':

        transitions = {
            'Up': ('S0', normal_reward),
            'Down': ('S1', normal_reward),
            'Left': ('S0', wall_reward),
            'Right': ('S0', wall_reward)
        }

    elif state == 'S1':

        transitions = {
            'Up': ('S0', normal_reward),
            'Down': ('S1', normal_reward),
            'Left': ('S1', wall_reward),
            'Right': ('G', goal_reward)
        }

    return transitions[action]


# 3. Initialize State Values

V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

print("INITIAL STATE VALUES")
print("V(S0) =", V['S0'])
print("V(S1) =", V['S1'])
print("V(G)  =", V['G'])


# 4. Calculate Action Values

def calculate_q(state, action, V):

    next_state, reward = transition(state, action)

    return reward + gamma * V[next_state]


# 5. Implement Value Iteration

iteration = 1

while True:

    delta = 0
    new_V = V.copy()

    print("\n" + "=" * 50)
    print("ITERATION", iteration)
    print("=" * 50)

    for state in states:

        q_values = {}

        for action in actions:

            q_values[action] = calculate_q(state, action, V)

        best_value = max(q_values.values())

        new_V[state] = best_value

        print("\nState:", state)

        for action in actions:
            print(
                f"Q({state}, {action}) = "
                f"{q_values[action]:.2f}"
            )

        print("Updated Value =", round(best_value, 4))

        delta = max(delta, abs(new_V[state] - V[state]))

    V.update(new_V)

    print("\nSTATE VALUE TABLE")

    for state in states:
        print(f"V({state}) = {V[state]:.4f}")

    print("Maximum Change =", round(delta, 6))

    if delta < theta:
        print("\nValue Iteration Converged!")
        break

    iteration += 1


# 6. Calculate Final Optimal Action Values

print("\n" + "=" * 50)
print("FINAL OPTIMAL ACTION VALUES")
print("=" * 50)

for state in states:

    print("\nState:", state)

    for action in actions:

        q = calculate_q(state, action, V)

        print(f"Q*({state}, {action}) = {q:.2f}")


# 7. Find Optimal Policy

print("\n" + "=" * 50)
print("OPTIMAL POLICY")
print("=" * 50)

policy = {}

for state in states:

    q_values = {}

    for action in actions:

        q_values[action] = calculate_q(state, action, V)

    best_action = max(q_values, key=q_values.get)

    policy[state] = best_action

    print(f"{state} -> {best_action}")

print("G -> Terminal")


# 8. Display Final Optimal State Values

print("\n" + "=" * 50)
print("FINAL OPTIMAL STATE VALUES")
print("=" * 50)

for state in states:
    print(f"V*({state}) = {V[state]:.2f}")

print("V*(G) = 0.00")


# 9. Bellman Optimality Verification

print("\n" + "=" * 50)
print("BELLMAN OPTIMALITY VERIFICATION")
print("=" * 50)

for state in states:

    q_values = []

    for action in actions:

        q = calculate_q(state, action, V)

        q_values.append(q)

    max_q = max(q_values)

    print(f"\nV*({state}) = {V[state]:.2f}")
    print(f"Max Q*({state}, a) = {max_q:.2f}")

    if abs(V[state] - max_q) < 0.001:
        print("Bellman Equation Verified!")


# 10. Display Optimal Path

print("\n" + "=" * 50)
print("OPTIMAL PATH")
print("=" * 50)

print("S0 -> S1 -> G")
```
